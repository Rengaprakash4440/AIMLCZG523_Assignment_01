# Heart Disease UCI - Exploratory Data Analysis
Run from the `notebooks/` folder after `python src/download_data.py` and creating `data/heart_clean.csv`.

```python
import sys
from pathlib import Path

ROOT = Path.cwd()
while not (ROOT / "src" / "data.py").exists() and ROOT != ROOT.parent:
    ROOT = ROOT.parent
assert (ROOT / "src" / "data.py").exists(), f"src/data.py not found from {Path.cwd()}"
sys.path.insert(0, str(ROOT))

import pandas as pd, matplotlib.pyplot as plt, seaborn as sns
from src.data import NUMERIC, CATEGORICAL

sns.set_theme(style="whitegrid")
FIG = ROOT / "report" / "figures"; FIG.mkdir(parents=True, exist_ok=True)
df = pd.read_csv(ROOT / "data" / "heart_clean.csv")
df.head()
```

## 1. Overview and missing values

```python
print(df.shape)
display(df.describe().T)
display(df.isna().sum().to_frame("missing").query("missing > 0"))
```

## 2. Class balance

```python
ax = sns.countplot(x="target", data=df)
ax.set_title("Class balance (0 = no disease, 1 = disease)")
for c in ax.containers: ax.bar_label(c)
plt.savefig(FIG / "class_balance.png", dpi=150, bbox_inches="tight"); plt.show()
```

## 3. Numeric feature distributions (by target)

```python
fig, axes = plt.subplots(2, 3, figsize=(15, 8))
for ax, col in zip(axes.ravel(), NUMERIC):
    sns.histplot(data=df, x=col, hue="target", kde=True, ax=ax)
axes.ravel()[-1].axis("off")
plt.tight_layout()
plt.savefig(FIG / "numeric_histograms.png", dpi=150); plt.show()
```

## 4. Boxplots vs target (outliers)

```python
fig, axes = plt.subplots(1, 5, figsize=(18, 4))
for ax, col in zip(axes, NUMERIC):
    sns.boxplot(data=df, x="target", y=col, ax=ax)
plt.tight_layout()
plt.savefig(FIG / "numeric_boxplots.png", dpi=150); plt.show()
```

## 5. Categorical features vs target

```python
fig, axes = plt.subplots(2, 4, figsize=(18, 8))
for ax, col in zip(axes.ravel(), CATEGORICAL):
    sns.countplot(data=df, x=col, hue="target", ax=ax)
plt.tight_layout()
plt.savefig(FIG / "categorical_vs_target.png", dpi=150); plt.show()
```

## 6. Correlation heatmap

```python
plt.figure(figsize=(10, 8))
sns.heatmap(df.corr(numeric_only=True), annot=True, fmt=".2f", cmap="coolwarm", center=0)
plt.title("Correlation heatmap")
plt.savefig(FIG / "correlation_heatmap.png", dpi=150, bbox_inches="tight"); plt.show()
```
